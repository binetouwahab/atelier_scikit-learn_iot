Partie 2 – Sélection de y (cible) et X (caractéristiques)
1) Définir "etat" comme la cible ou valeur à prédire et "temperature", "humidite",
"pression" et "consommation" comme caractéristiques ou variables explicatives


```python
df = df.dropna(subset=["etat"])
```

```python

X = df[["temperature", "humidite", "pression", "consommation"]]   #les caracteristiques
y = df["etat"]   #le cible
```

3) Quel est le type du problème de machine learning ?

C'est un problème de classification supervisée parce qu'on connaît déjà la valeur etat dans les données d'entraînement et on veut apprendre à la prédire à partir de temperature, humidite, pression et consommation.

Partie 3 – Découpage Train/Test
Diviser X en deux ensembles distincts : un pour l'entraînement (train) et un pour le test (test).
Avec les conditions suivantes : 20% des données serviront au test ; garantir la reproductibilité du
découpage ; conserver les mêmes proportions de classes dans l'ensemble de train et de test que
dans les données d'origine.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2, #20 % des données pour le test et 80 % pour l'entraînement.
    random_state=42, #garantit que le découpage sera toujours le même.
    stratify=y #conserve les mêmes proportions de classes dans y_train et y_test
)
```

Partie 4 – Gestion des valeurs manquantes
1) Vérifier l’existence de valeurs manquantes

```python
print(X_train.isna().sum())
print("_____________________")
print(X_test.isna().sum())
```

2) Sélectionner SimpleImputer avec la médiane

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
```

3) Qu’est ce qui justifie le choix de la médiane ?

La médiane a été choisie car les variables numériques contiennent des valeurs extrêmes. Contrairement à la moyenne, la médiane est peu sensible aux valeurs aberrantes et permet donc une imputation plus robuste.

4) Trouver les paramètres (médianes) de l’imputeur sur X_train

```python
imputer.fit(X_train) #Calcule les médianes uniquement à partir des données d'entraînement.
```

```python
print(imputer.statistics_) #pour voir les médianes calculées
```

5) Déterminer X_train_imputed et X_test_imputed, les transformés de X_train et X_test

```python
X_train_imputed = imputer.transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
print(pd.DataFrame(X_train_imputed).isna().sum())
print(pd.DataFrame(X_test_imputed).isna().sum())
```

1) Sélectionner StandardScaler pour mettre à l’échelle les transformés de l’imputation

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(X_train_imputed)
```

2) Qu’est ce qui justifie la standardisation ?

La standardisation transforme chaque variable pour qu'elle soit approximativement :
moyenne = 0
écart-type = 1

3) Trouver les paramètres (moyennes et écart-types) du scaleur sur X_train_imputed

```python
print("Moyennes :", scaler.mean_)
print("Écarts-types :", scaler.scale_)
```

4) Déterminer X_train_scaled et X_test_scaled, les transformés de X_train_imputed et
X_test_imputed

```python
scaler.fit(X_train_imputed)

X_train_scaled = scaler.transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

Partie 6 – Entrainement et prédiction d’un modèle
1) Sélectionner le modèle KNN (k plus proches voisins) avec 5, le nombre de voisins à prendre en compte

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
```

2) Entrainer le modèle

```python
knn.fit(X_train_scaled, y_train)
```

3) Déterminer y_pred, la prédiction du modèle avec l’ensemble de test

```python
y_pred = knn.predict(X_test_scaled)
print(y_pred)
```

4) Afficher quelques prédictions

```python
print(y_pred[:10])
```

5) Comparer ces prédictions avec les vraies valeurs

```python
print("Prédictions :", y_pred[:10])
print("Vraies valeurs :", y_test[:10])
```

Partie 7 – Evaluation du modèle
1) Evaluer le modèle avec Accuracy puis afficher le résultat

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Accuracy en pourcentage :", accuracy * 100, "%")
```


2) Pourquoi Accuracy ne suffit pas toujours pour évaluer ?

L'Accuracy peut être trompeuse lorsque les classes sont déséquilibrées

3) Déterminer, afficher et commenter la matrice de confusion

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

Classe 0 : il y avait 6 observations réelles → 2 correctement prédites comme 0, mais 4 prédites comme 2.
Classe 1 : il y avait 1 observation réelle → elle a été prédite comme 0. Donc 0 bonne prédiction pour cette classe.
Classe 2 : il y avait 113 observations réelles → les 113 ont été correctement prédites comme 2.

La matrice de confusion montre que le modèle classe correctement les 113 observations de la classe 2. En revanche, il rencontre des difficultés pour les classes 0 et 1 : sur les 6 observations de la classe 0, seulement 2 sont correctement classées, tandis que l'unique observation de la classe 1 est mal classée. Le modèle semble donc fortement favoriser la classe 2, ce qui peut s'expliquer par un déséquilibre important entre les classes.

4) Visualiser cette matrice avec Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(cm, annot=True, fmt="d")

plt.xlabel("Prédictions")
plt.ylabel("Vraies valeurs")
plt.title("Matrice de confusion")

plt.show()
```

5) Afficher et commenter le rapport de classificaton

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

Le rapport de classification montre que le modèle obtient de très bonnes performances pour la classe 2, qui est largement majoritaire. En revanche, les performances sont faibles pour les classes 0 et 1, notamment pour la classe 1 qui n'est pas correctement prédite. Ces résultats s'expliquent notamment par le fort déséquilibre entre les classes.
